### **First and Second Order Methods**

**Gradient Descent**

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
b = 0.001
eta = 0.2
iterations = 50

# Initial values
x = np.array([2.0, 2.0, 2.0])

# History arrays for plotting
x1_hist, x2_hist, x3_hist = [x[0]], [x[1]], [x[2]]

for _ in range(iterations):
    # Calculate gradients
    grad_x1 = x[0]**3
    grad_x2 = b * (x[1]**2)
    grad_x3 = (2 * b * x[2]) / (1 + x[2]**2)
    
    # Update variables
    x[0] = x[0] - eta * grad_x1
    x[1] = x[1] - eta * grad_x2
    x[2] = x[2] - eta * grad_x3
    
    # Store history
    x1_hist.append(x[0])
    x2_hist.append(x[1])
    x3_hist.append(x[2])

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(x1_hist, label='x1', marker='o', markersize=4)
plt.plot(x2_hist, label='x2', marker='s', markersize=4)
plt.plot(x3_hist, label='x3', marker='^', markersize=4)

plt.title('Change of x1, x2, x3 over 50 Iterations')
plt.xlabel('Iteration Number')
plt.ylabel('Parameter Value')
plt.legend()
plt.grid(True)
plt.show()
```
